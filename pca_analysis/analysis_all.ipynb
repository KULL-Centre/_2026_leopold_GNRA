```python
import pickle
import os
import numpy as np
```

```python
with open('pca_results.pkl', 'rb') as f:
    # Load the PCA results from the file
    pca_results = pickle.load(f)
```

```python
w=pca_results['w']
v=pca_results['v']
```

```python
names=pca_results['names']

pdb_names = []
for name in names:
    _, _, pdb_name = name.partition('/')
    pdb_names.append(pdb_name)

clusters_w = {}
for i in range(1, 45):
    if i < 10:
        cluster_name = '0' + str(i)
    else:
        cluster_name = str(i)
    cluster_w = []
    for name,pca_w in zip(pdb_names,w):
        if os.path.isfile(f'cluster/C{cluster_name}/{name}') == True:
            print(f'cluster/C{cluster_name}/{name} found')
            cluster_w.append(pca_w)
        else:
            continue
    cluster_w = np.array(cluster_w)
    assert len(cluster_w) > 0, f"No data found for cluster {i}"
    clusters_w[f'cluster_{i}'] = cluster_w
```

```python
for i in range(0,7):
    cluster_w_gcaa = []
    for name,pca_w in zip(pdb_names,w):
        if f'cluster_{i}_gcaa_' in name:
            print(f'{name} found for gcaa')
            cluster_w_gcaa.append(pca_w)
        else:
            continue
    assert len(cluster_w_gcaa) > 0, "No data found for gcaa"
    clusters_w[f'cluster_{i}_gcaa'] = np.array(cluster_w_gcaa)

for i in range(0,5):
    cluster_w_gaag = []
    for name,pca_w in zip(pdb_names,w):
        if f'cluster_{i}_gaag_' in name:
            print(f'{name} found for gaag')
            cluster_w_gaag.append(pca_w)
        else:
            continue
    assert len(cluster_w_gaag) > 0, "No data found for gaag"
    clusters_w[f'cluster_{i}_gaag'] = np.array(cluster_w_gaag)
```

```python
ensemble_types = ['gaag_aria','gaag_Farfar1', 'gaag_Farfar2', 'gaag_alphafold3', 'gcaa_aria_1', 'gcaa_aria_2','gcaa_Farfar1', 'gcaa_Farfar2', 'gcaa_alphafold3']

for ensemble_type in ensemble_types:
    cluster_w = []
    for name,pca_w in zip(pdb_names,w):
        if ensemble_type in name:
            print(f'{name} found for {ensemble_type}')
            cluster_w.append(pca_w)
        else:
            continue
    assert len(cluster_w) > 0, f"No data found for {ensemble_type}"
    clusters_w[f'cluster_{ensemble_type}'] = np.array(cluster_w)
```

```python
with open('clusters_w_pca.pkl', 'wb') as f:
    pickle.dump(clusters_w, f)
```

```python
with open('clusters_w_pca.pkl', 'rb') as f:
    # Load the PCA results from the file
    clusters_w = pickle.load(f)
```

```python
print(clusters_w.keys())
```

```python
cluster_plot_assignments = {
    'cluster_1': 'GNRA',
    'cluster_2': 'UNCG',
    'cluster_4': 'U-TURN',
    'cluster_11': '4 STACK',
    'cluster_21': 'RNYA',
    'cluster_30':'GGUG',
    'cluster_35':'CUUG',
    'cluster_38':'CUUG',
    'cluster_39':'AGNN'
}

cluster_color_assignments = {
    'GNRA':'red',
    'UNCG':'cornflowerblue',
    'U-TURN':'green',
    '4 STACK':'purple',
    'RNYA':'orange',
    'GGUG':'gold',
    'CUUG':'brown',
    'AGNN':'hotpink'
}
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("white")

plt.scatter(w[:,0],w[:,1])
plt.xlabel("PC1")
plt.ylabel("PC2")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("white")

plt.scatter(w[:,0],w[:,2])
plt.xlabel("PC1")
plt.ylabel("PC3")
```

```python
fig = plt.figure(figsize=(5,5))
tloop = "gaag"
plt.title(f"{tloop.upper()} ensemble clusters")


plt.scatter(w[:,0],w[:,1], c='gray', alpha=0.2, s=5)
for cluster_name in cluster_plot_assignments :
    cluster_w = clusters_w[cluster_name]
    plt.scatter(cluster_w[:,0],cluster_w[:,1], s=10, c=cluster_color_assignments[cluster_plot_assignments[cluster_name]], label=cluster_plot_assignments[cluster_name])

for i in range(0, 5):
    cluster_w = clusters_w[f'cluster_{i}_{tloop}']
    plt.scatter(cluster_w[:,0],cluster_w[:,1], s=20, edgecolor='k', label=f'{tloop.upper()} cluster {i}')

plt.xlabel("PC1")
plt.ylabel("PC2")

plt.xlim(max(w[:,0]), min(w[:,0]))
plt.ylim(max(w[:,1]), min(w[:,1])) 

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)

plt.savefig(f'figures/{tloop}_pca_PC1_PC2_clusters.png', dpi=300, bbox_inches='tight')
plt.savefig(f'figures/{tloop}_pca_PC1_PC2_clusters.pdf', dpi=300, bbox_inches='tight')
```
